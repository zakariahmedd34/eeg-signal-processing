# Creating epochs and generating evoked response (ERP/ERF)

```python
import matplotlib
import pathlib

import mne
import mne_bids

matplotlib.use('QtAgg')
mne.set_config("MNE_BROWSER_THEME",  "light")
```

```python
bids_root = pathlib.Path('out_data/sample_BIDS')

bids_path = mne_bids.BIDSPath(
    subject='01',
    session='01',
    task="audiovisual",
    run='01',
    datatype='meg',
    root=bids_root
)

raw = mne_bids.read_raw_bids(bids_path)
raw.load_data()
raw.filter(l_freq=0.1,h_freq=40)
events, event_id = mne.events_from_annotations(raw)
```

```python
raw.plot()
```

```python
event_id
```

```python
tmin = -0.3
tmax =  0.5
baseline = (None, 0)

epochs = mne.Epochs(
    raw=raw,
    events=events,
    event_id=event_id,
    tmin=tmin,
    tmax=tmax,
    baseline=baseline,
    preload=True
)
epochs
```

```python
epochs.plot()
```

```python
epochs
```

```python
e_min = -0.25
e_max = 0.80
baseline = (-0.20,0)

epochs = mne.Epochs(
    raw=raw,
    events=events,
    event_id = event_id,
    tmin = e_min,
    tmax= e_max,
    baseline = baseline,
    preload=True
     
)
epochs
```

```python
epochs.plot()
```
